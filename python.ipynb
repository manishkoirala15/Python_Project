```python

#the __init__()
class Student:
  def __init__(self, name, age):#self means the current object itself
    self.name = name
    self.age = age
s1 = Student("Babin",20)
s2 = Student("Bishnu",52)
print(s1.name, s2.age)
print(s2.name, s1.age)
```
